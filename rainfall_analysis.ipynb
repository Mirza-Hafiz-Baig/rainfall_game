```python
df = pd.read_csv("rainfall in india 1901-2015.csv")

df.head()
```

```python
months = [
    "JAN", "FEB", "MAR", "APR", "MAY", "JUN",
    "JUL", "AUG", "SEP", "OCT", "NOV", "DEC"
]

df.columns = df.columns.str.strip()
df["SUBDIVISION"] = df["SUBDIVISION"].str.strip()

# Calculate total rainfall for each year
df["ANNUAL_RAINFALL"] = df[months].sum(axis=1, min_count=1)

# Average annual rainfall for each region
regional_rainfall = (
    df.groupby("SUBDIVISION")["ANNUAL_RAINFALL"]
      .mean()
      .sort_values(ascending=False)
)

print("Top 10 regions by average annual rainfall:")
display(regional_rainfall.head(10).to_frame())

print("\nOverall average annual rainfall:")
print(round(df["ANNUAL_RAINFALL"].mean(), 2), "mm")
```
